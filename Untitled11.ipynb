```python
import pandas as pd
import zipfile
import requests

url = "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip"

response = requests.get(url)
with open("movies.zip", "wb") as f:
    f.write(response.content)

with zipfile.ZipFile("movies.zip", "r") as zip_ref:
    zip_ref.extractall(".")

movies = pd.read_csv("ml-latest-small/movies.csv")
ratings = pd.read_csv("ml-latest-small/ratings.csv")

print(movies.head())
```

```python
movies["genres"] = movies["genres"].str.replace("|", " ", regex=False)

movies["content"] = (
    movies["title"] + " " + movies["genres"]
)

print(movies[["title", "genres"]].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(movies["content"])

similarity = cosine_similarity(tfidf_matrix)

def recommend_movies(movie_name, top_n=5):

    matches = movies[
        movies["title"].str.contains(
            movie_name, case=False, regex=False
        )
    ]

    if matches.empty:
        return "Movie not found."

    index = matches.index[0]

    scores = list(enumerate(similarity[index]))

    scores = sorted(
        scores, key=lambda x: x[1], reverse=True
    )

    results = []

    for i, score in scores[1:top_n+1]:
        results.append({
            "Movie": movies.iloc[i]["title"],
            "Genres": movies.iloc[i]["genres"],
            "Similarity": round(score, 3)
        })

    return pd.DataFrame(results)

recommend_movies("Toy Story")
```

```python
from transformers import pipeline

llm = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)

def understand_preferences(user_input):

    prompt = f"""
    Extract the movie preferences from this request.
    Return the preferred genres and movie characteristics.

    User request: {user_input}
    """

    result = llm(prompt, max_new_tokens=100)

    return result[0]["generated_text"]

understand_preferences(
    "I want exciting space science fiction movies"
)
```

```python
from functools import lru_cache

@lru_cache(maxsize=1000)
def get_tmdb_movie(title):

    import re

    year_match = re.search(r"\((\d{4})\)", title)
    year = year_match.group(1) if year_match else None

    clean_title = re.sub(r"\s*\(\d{4}\)\s*$", "", title)

    params = {"query": clean_title}

    if year:
        params["year"] = year

    response = requests.get(
        "https://api.themoviedb.org/3/search/movie",
        headers=headers,
        params=params,
        timeout=20
    )

    response.raise_for_status()

    results = response.json().get("results", [])

    if not results and year:
        params.pop("year")
        response = requests.get(
            "https://api.themoviedb.org/3/search/movie",
            headers=headers,
            params=params,
            timeout=20
        )
        response.raise_for_status()
        results = response.json().get("results", [])

    if not results:
        return {}

    movie = results[0]

    details_response = requests.get(
        f"https://api.themoviedb.org/3/movie/{movie['id']}",
        headers=headers,
        timeout=20
    )

    details_response.raise_for_status()
    details = details_response.json()

    credits_response = requests.get(
        f"https://api.themoviedb.org/3/movie/{movie['id']}/credits",
        headers=headers,
        timeout=20
    )

    credits = credits_response.json()

    directors = [
        p["name"] for p in credits.get("crew", [])
        if p.get("job") == "Director"
    ]

    return {
        "id": movie["id"],
        "poster": (
            TMDB_IMAGE + movie["poster_path"]
            if movie.get("poster_path") else ""
        ),
        "overview": details.get("overview", ""),
        "rating": details.get("vote_average", 0),
        "director": ", ".join(directors) or "Not available",
        "year": details.get("release_date", "")[:4]
    }
```

```python

import gradio as gr
import pandas as pd
import numpy as np
import html
import re
from sklearn.metrics.pairwise import cosine_similarity

# ==========================================
# 1. PREPARE MOVIE DATA
# ==========================================

# Make sure movie IDs and ratings are available
rating_stats = ratings.groupby("movieId").agg(
    mean_rating=("rating", "mean"),
    rating_count=("rating", "count")
).reset_index()

movie_data = movies.merge(
    rating_stats,
    on="movieId",
    how="left"
)

movie_data["mean_rating"] = movie_data["mean_rating"].fillna(0)
movie_data["rating_count"] = movie_data["rating_count"].fillna(0)

# ==========================================
# 2. RECOMMENDATION ENGINE
# ==========================================

def get_movie_results(query="", genre="All", sort="Popular"):

    if query.strip():

        # First check whether user entered a movie title
        matches = movies[
            movies["title"].str.contains(
                query.strip(), case=False, regex=False
            )
        ]

        if not matches.empty:
            movie_index = matches.index[0]
            scores = similarity[movie_index].copy()
            scores[movie_index] = -1

        else:
            # LLM understands natural language preferences
            try:
                preferences = understand_preferences(query)
            except Exception:
                preferences = query

            search_text = query + " " + preferences
            vector = tfidf.transform([search_text])

            scores = cosine_similarity(
                vector, tfidf_matrix
            ).flatten()

        indices = np.argsort(scores)[::-1]

    else:
        # Popular movies from MovieLens
        indices = movie_data.sort_values(
            ["rating_count", "mean_rating"],
            ascending=False
        ).index.to_numpy()

    output = []

    for idx in indices:

        row = movie_data.iloc[idx]

        title = str(row["title"])
        genres = str(row["genres"])

        if genre != "All" and genre not in genres:
            continue

        # Fetch real TMDB poster and movie details
        info = get_tmdb_movie(title)

        if not info or not info.get("poster"):
            continue

        if sort == "Top Rated":
            if float(info.get("rating", 0)) < 6:
                continue

        if sort == "Newest":
            year = info.get("year", "")
            if not year or int(year) < 2010:
                continue

        output.append({
            "title": title,
            "genres": genres,
            "info": info
        })

        if len(output) >= 20:
            break

    return output


# ==========================================
# 3. CREATE MOVIE CARDS
# ==========================================

def render_cards(movie_list):

    cards = []

    for movie in movie_list:

        title = html.escape(movie["title"])
        genres = html.escape(movie["genres"])

        info = movie["info"]

        poster = info.get("poster", "")
        overview = html.escape(
            info.get("overview") or "No description available."
        )

        director = html.escape(
            info.get("director", "Not available")
        )

        rating = float(info.get("rating", 0))
        year = html.escape(info.get("year", ""))

        movie_id = info.get("id")

        poster_html = f"""
        <img src="{poster}"
             class="poster"
             loading="lazy"
             alt="{title}"
             onerror="this.style.display='none';
             this.nextElementSibling.style.display='flex';">
        <div class="poster-fallback" style="display:none;">
            Poster unavailable
        </div>
        """

        movie_url = (
            f"https://www.themoviedb.org/movie/{movie_id}"
            if movie_id else "https://www.themoviedb.org/"
        )

        cards.append(f"""
        <div class="movie-card">

            <div class="poster-box">

                {poster_html}

                <div class="rating">
                    ⭐ {rating:.1f}/10
                </div>

            </div>

            <div class="movie-info">

                <h3>{title}</h3>

                <div class="genres">
                    {year} | {genres}
                </div>

                <div class="director">
                    Director: <span>{director}</span>
                </div>

                <p class="description">
                    {overview}
                </p>

                <div class="actions">

                    <a href="{movie_url}"
                       target="_blank"
                       class="play-button">
                       ▶ DETAILS
                    </a>

                    <span class="heart">♡</span>

                </div>

            </div>
        </div>
        """)

    return f"""
    <div class="results-heading">
        <h2>🎬 Recommended Movies</h2>
        <span>{len(cards)} movies found</span>
    </div>

    <div class="movie-grid">
        {''.join(cards)}
    </div>
    """


# ==========================================
# 4. SEARCH HANDLER
# ==========================================

def search_movies(query, genre, sort):

    try:
        results = get_movie_results(query, genre, sort)

        if not results:
            return """
            <div class="empty">
                <h2>🎬 No movies found</h2>
                <p>Try another search or select a different genre.</p>
            </div>
            """

        return render_cards(results)

    except Exception as e:
        return f"""
        <div class="empty">
            <h3>Error loading recommendations</h3>
            <p>{html.escape(str(e))}</p>
        </div>
        """


# ==========================================
# 5. PROFESSIONAL CINEMATE CSS
# ==========================================

custom_css = """

.gradio-container {
    background: #101417 !important;
    color: #ffffff !important;
    max-width: 1600px !important;
}

#header {
    background: #191f23;
    border: 1px solid #30383e;
    border-radius: 10px;
    padding: 12px 20px;
}

.logo {
    color: #f5a623;
    font-size: 26px;
    font-weight: 900;
}

.logo span {
    color: white;
}

#sidebar {
    background: #202930;
    border-radius: 9px;
    padding: 15px;
    min-height: 700px;
}

.sidebar-title {
    color: #f5a623;
    font-size: 13px;
    font-weight: bold;
    margin: 20px 0 12px;
}

.sidebar-item {
    padding: 10px;
    color: #c9d0d4;
    font-size: 13px;
    border-radius: 5px;
    margin-bottom: 5px;
}

.sidebar-item:hover {
    background: #343d43;
}

#discover-title h1 {
    font-size: 27px;
    color: #f5a623;
}

.results-heading {
    display: flex;
    justify-content: space-between;
    align-items: center;
    margin: 20px 0;
}

.results-heading h2 {
    color: #f5a623;
    font-size: 21px;
}

.results-heading span {
    color: #aeb8bf;
    font-size: 12px;
}

.movie-grid {
    display: grid;
    grid-template-columns: repeat(4, minmax(0, 1fr));
    gap: 15px;
}

.movie-card {
    background: #191f23;
    border: 1px solid #343d43;
    border-radius: 11px;
    overflow: hidden;
    transition: all .25s ease;
    min-width: 0;
}

.movie-card:hover {
    transform: translateY(-5px);
    border-color: #e6a000;
    box-shadow: 0 7px 20px #0007;
}

.poster-box {
    position: relative;
    width: 100%;
    height: 255px;
    background: #252d32;
}

.poster {
    width: 100%;
    height: 100%;
    object-fit: cover;
    display: block;
}

.poster-fallback {
    height: 100%;
    align-items: center;
    justify-content: center;
    color: #c7c7c7;
}

.rating {
    position: absolute;
    bottom: 9px;
    right: 8px;
    background: #080a0b;
    color: #fff;
    padding: 6px 9px;
    border-radius: 5px;
    font-size: 11px;
}

.movie-info {
    padding: 12px;
}

.movie-info h3 {
    font-size: 14px;
    color: #fff;
    min-height: 34px;
    margin: 0 0 7px;
}

.genres {
    color: #f5a623;
    font-size: 11px;
    min-height: 25px;
}

.director {
    color: #ddd;
    font-size: 11px;
    margin-top: 9px;
}

.director span {
    color: #e6a623;
}

.description {
    font-size: 11px;
    line-height: 1.5;
    color: #aab3b9;
    height: 49px;
    overflow: hidden;
    margin-top: 10px;
}

.actions {
    display: flex;
    justify-content: space-between;
    align-items: center;
    margin-top: 13px;
}

.play-button {
    color: #ffb000 !important;
    border: 1px solid #e5a000;
    padding: 8px 13px;
    border-radius: 5px;
    text-decoration: none !important;
    font-size: 11px;
    font-weight: bold;
}

.heart {
    background: #30393e;
    padding: 7px 12px;
    border-radius: 6px;
    color: #ffb000;
    font-size: 20px;
}

.empty {
    text-align: center;
    padding: 70px 10px;
    color: #f5a623;
}

#search-btn {
    background: #efa900;
    color: #111;
    font-weight: bold;
    border: none;
}

@media(max-width: 1000px) {
    .movie-grid {
        grid-template-columns: repeat(3, minmax(0, 1fr));
    }
}

@media(max-width: 700px) {
    .movie-grid {
        grid-template-columns: repeat(2, minmax(0, 1fr));
    }
}

@media(max-width: 450px) {
    .movie-grid {
        grid-template-columns: 1fr;
    }
}

footer {
    display: none !important;
}

"""


# ==========================================
# 6. BUILD GRADIO APPLICATION
# ==========================================

with gr.Blocks(
    theme=gr.themes.Base(
        primary_hue="amber",
        neutral_hue="slate"
    ),
    css=custom_css,
    title="Cinemate AI"
) as demo:

    # TOP NAVIGATION

    with gr.Row(elem_id="header", equal_height=True):

        gr.HTML("""
        <div class="logo">🎞 Cinemate</div>
        """)

        search_input = gr.Textbox(
            placeholder="🔍 Search movies or describe your taste...",
            show_label=False,
            container=False,
            scale=5
        )

        search_button = gr.Button(
            "SEARCH",
            elem_id="search-btn",
            scale=1
        )

    with gr.Row():

        # SIDEBAR

        with gr.Column(scale=1, elem_id="sidebar"):

            gr.HTML("""
            <div class="sidebar-title">DISCOVER</div>
            <div class="sidebar-item">🎬 All Movies</div>
            <div class="sidebar-item">🔥 Popular Movies</div>
            <div class="sidebar-item">⭐ Top Rated</div>

            <div class="sidebar-title">RECOMMENDATIONS</div>
            <div class="sidebar-item">✨ AI Recommendations</div>
            <div class="sidebar-item">❤️ Favourite Movies</div>

            <div class="sidebar-title">ABOUT</div>
            <div class="sidebar-item">🤖 AI Powered</div>
            <div class="sidebar-item">🎥 TMDB Movies</div>
            """)

            genre_select = gr.Dropdown(
                choices=[
                    "All", "Action", "Adventure", "Animation",
                    "Comedy", "Crime", "Documentary", "Drama",
                    "Fantasy", "Horror", "Romance", "Sci-Fi",
                    "Thriller", "War", "Western"
                ],
                value="All",
                label="Genres"
            )

            sort_select = gr.Dropdown(
                choices=["Popular", "Top Rated", "Newest"],
                value="Popular",
                label="Sort Movies"
            )

        # MAIN CONTENT

        with gr.Column(scale=5):

            gr.HTML("""
            <div id="discover-title">
                <h1>Discover Movies 🍿</h1>
                <p style="color:#aeb8bf;font-size:13px;">
                Find your next favourite movie with AI.
                </p>
            </div>
            """)

            movie_output = gr.HTML()

            gr.Examples(
                examples=[
                    ["Interstellar"],
                    ["Inception"],
                    ["I love science fiction and space movies"],
                    ["Recommend action and adventure movies"],
                    ["I enjoy romantic comedy movies"]
                ],
                inputs=search_input
            )

    # EVENTS

    search_button.click(
        fn=search_movies,
        inputs=[
            search_input,
            genre_select,
            sort_select
        ],
        outputs=movie_output
    )

    search_input.submit(
        fn=search_movies,
        inputs=[
            search_input,
            genre_select,
            sort_select
        ],
        outputs=movie_output
    )

    genre_select.change(
        fn=search_movies,
        inputs=[
            search_input,
            genre_select,
            sort_select
        ],
        outputs=movie_output
    )

    sort_select.change(
        fn=search_movies,
        inputs=[
            search_input,
            genre_select,
            sort_select
        ],
        outputs=movie_output
    )

    # INITIAL MOVIES

    demo.load(
        fn=search_movies,
        inputs=[
            search_input,
            genre_select,
            sort_select
        ],
        outputs=movie_output
    )


# ==========================================
# 7. LAUNCH APPLICATION
# ==========================================

demo.launch(share=True)
```
